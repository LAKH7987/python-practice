```python
# company 
class Company :
    def __init__(self,name,sal,dept):
        self.name = name
        self.sal = sal
        self.dept = dept

    
    def salary(self) :
        print(f"salary of employee is : {self.sal}")
    
    def Bonus(self) :
        bonus = self.sal*0.05
        total = self.sal + bonus
        print(f" 5% bonus add in salary then total salary is :{total}\n ")

emp1 = Company("ankit",50000,"it")
emp1.salary()
emp1.Bonus()

emp2 = Company("Mahesh",45000,"MBBS")
emp2.salary()
emp2.Bonus()
```

```python
# class: animal ==> name, age, sound

class animal : 
    def __init__(self,name,age,sound):
        self.name = name
        self.age = age
        self.sound = sound
        print("information of animal")
    def show_info(self) :
        print(f"name : {self.name}")
        print(f"age is : {self.age}")
        print(f"sound is : {self.sound}\n") 

cat = animal("cat",3,"mew")
cat.show_info()

horse = animal("horse",6,"Neigh")
horse.show_info()

dog = animal("Dog",3,"Bark")
print(f"name : {dog.name}")
print(f"age : {dog.age}")
print(f"sound : {dog.sound}")
```

```python
# class: mobile ==> model_no. , review
class Mobile :
    def __init__(self,model_no,review) :
        self.model_no = model_no
        self.review = review
        print("mobile info.....")
    
    def show_info(self) :
        print(f"model number : {self.model_no}")
        print(f"reviews : {self.review}\n")

m1 = Mobile("galaxy s24",4.3)
m1.show_info()

m2 = Mobile("iphone 13 pro",4.7)
m2.show_info()
```

```python
# class: course ==> course name, course_code
class Course :
    def __init__(self,name,code) :
        self.name = name
        self.code = code
        print("course information..")
    
    def show_info(self) :
        print(f"course name : {self.name}")
        print(f"course code : {self.code}\n")

c1 = Course("Data Science","d101")
c1.show_info()

c2 = Course("Data Analytics","d101")
c2.show_info()
```

```python
# 5. Write a class BankAccount with attributes account_holder and balance. Add a method deposit(amount) that increases the balance.

class BankAccount :
    def __init__(self,acc_holder,balance) :
        self.acc_holder = acc_holder
        self.balance = balance

    def Deposit(self,amount = 0) :
        print(f"current bank balance : {self.balance}")
        deposit = self.balance + amount
        print(f"after deposit amount , bank balance : {deposit}")

b = BankAccount("mahesh",4000)
b.Deposit(1000)
```

```python
# 7. Write a class Rectangle that takes length and width in init and has a method area() that returns the area.

class Rectangle :
    def __init__(self,length,width) :
        self.length = length
        self.width = width

    def area(self) :
        return self.length * self.width

r = Rectangle(4,5)
print("area of rectangle is : ",r.area())
```
