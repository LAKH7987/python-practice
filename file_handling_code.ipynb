```python
# by using "with open" when we read any file and after not close then with open automatically close to the file after read
with open("file_handling.txt","r") as f :
    print(f.read(10))
    print(f.readline())
    print(f.readline())
```

```python
# by using loop read all lines of the file 
with open("file_handling.txt","r") as f :
   for i in f :
      print(i)
```

```python
with open("file_handling.txt","r") as f :
   for i in f :
      print(i.strip())    # strip ---> removes spaces between lines
```

```python
# write  ----> it used for write single string (sentences) in file
with open("file_handling.txt","w") as f :
    f.write("this is a new updated file\n")
```

```python
# writelines ---> writelines used for many lines in once 
with open("file_handling.txt","a") as f :
    f.writelines("in file added a new line \n this is a python file")
```

```python
# r+  (read + write)   --------->  r+ used for read and write in file firstly read and after write in file
file = open("file_handling.txt", "r+")

print(file.read())   

file.write("\nHello")  

file.close()
```

```python
with open("r+_write.txt","r") as file :
    print(file.read())
```

```python
# when we firstly write and after read in r+ then  changes in first line because cursor position is on first line 
file = open("r+_write.txt","r+")
file.write("hello")
file.read()        
```

```python
#w+    -------->   write + read
file = open("data.txt", "w+")

file.write("Hello Python")
file.seek(0)  
print(file.read())
file.close()
```

```python
#when in w+ firstly read and after write then show empty file because w+ firstly allow to write in file and after read 

file = open("data.txt", "w+")
file.write("this is a python programming")
file.seek(0)
file.read()
file.close()
```

```python
#a+ (append + read) -----> a+ used for append text in file and after read . if file doesn't exist then a+ create this file 
file = open("data.txt", "a+")

file.write("\nWelcome")

file.seek(0)

print(file.read())

file.close()
```
